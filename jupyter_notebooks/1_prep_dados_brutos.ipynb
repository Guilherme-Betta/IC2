```python
import pandas as pd
import numpy as np
import os
from openpyxl import load_workbook
from datetime import datetime
```

```python
### Carregamento do arquivo de interesse

dados_brutos = pd.read_excel(r'C:\_Master\programacao\Guilherme-Betta\IC2\data\in\dados_brutos.xlsx') #Faz o python ler o arquivo e transformá-lo num DataFrame
```

```python
### Impedir que o pandas registre o horário na coluna de datas
dados_brutos['data_normalizada'] = dados_brutos['Data'].dt.date   # Cria uma coluna que extrai apenas ano-mes-dia da coluna "Data"

col = dados_brutos.pop('data_normalizada')   # Extrai o value da coluna criada para a linha seguinte

dados_brutos.insert(1, 'data_normalizada', col)  # Move a coluna criada para o começo da tabela
```

```python
### Criar um arquivo de saida 

pasta_saida = r"C:\_Master\programacao\Guilherme-Betta\IC2\data\out" # Cria uma variável para guardar o caminho de onde o arquivo resultante será salvo

os.makedirs(pasta_saida, exist_ok=True)   # Cria um diretorio para salvar o arquivo caso ele ainda nao exista

nome_arquivo = f"dados_brutos_prontos_{datetime.now().strftime('%Y%m%d_%H%M%S')}.xlsx" # Gera o nome do arquivo de saida com timestamp de quando foi criado

caminho_saida = os.path.join(pasta_saida, nome_arquivo)  # Cria o diretório para salvar o arquivo caso ela ainda não exista

dados_brutos.to_excel(caminho_saida, engine='openpyxl', index=True)  # Gera o arquivo de saida

print(f"Arquivo salvo em: {caminho_saida}") # Gera uma mensagem para o usuario informando sobre onde o arquivo foi salvo
```
