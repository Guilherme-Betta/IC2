```python
import pandas as pd
import os
from datetime import datetime
```

```python
arquivo_entrada = "C:\_Master\programacao\Guilherme-Betta\IC2\data\in\dadosBrutosProntos.xlsx"

dados = pd.read_excel(arquivo_entrada)

metadados = ['Data', 'Estatistica', 'data_normalizada']

variaveis = [cols for cols in dados.columns if cols not in metadados]

dados.head()
```

# Remoção de dados corrompidos (não numéricos)

```python
### Dados Corrompidos

for coluna in dados[variaveis]:
    dados[coluna] = pd.to_numeric(dados[coluna], errors='coerce')   # Remove valores de dtype != float das colunas na lista "variaveis"
    
dados.select_dtypes(exclude=float)
```

# Remoção de colunas com 30% de dados faltantes

```python
limpeza = dados.dropna(axis=1, thresh=len(dados)*0.3)   # Elimina colunas com metade (ou mais) de dados faltantes

limpeza.head()
```

# Salvamento de arquivo

```python
pasta_saida = "C:\_Master\programacao\Guilherme-Betta\IC2\data\out" # Cria uma variável para guardar o caminho de onde o arquivo resultante será salvo

os.makedirs(pasta_saida, exist_ok=True) # Passo necessário para o python salvar o arquivo

nome_arquivo = f"describe()_{datetime.now().strftime('%Y%m%d_%H%M%S')}.xlsx"   # Gera o nome do arquivo de saida com timestamp de quando foi criado

caminho_saida = os.path.join(pasta_saida, nome_arquivo) # Cria o diretório para salvar o arquivo caso ela ainda não exista

limpeza.to_excel(caminho_saida, engine='openpyxl', index=True)   # Gera o arquivo de saida
```
