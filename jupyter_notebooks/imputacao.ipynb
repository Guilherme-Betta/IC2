```python
import os
import numpy as np
import pandas as pd
import sklearn
from datetime import datetime
```

```python
arquivo_entrada = r"C:\_Master\programacao\Guilherme-Betta\IC2\data\out\limpeza.xlsx"

dados = pd.read_excel(arquivo_entrada)

metadados = ['Data', 'Estatistica', 'data_normalizada']

variaveis = [coluna for coluna in dados.select_dtypes(include=float)]
```

# KNN

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=2, weights="uniform")

X = dados[variaveis]

dados[variaveis] = imputer.fit_transform(X)
```

```python
# Verificar se ainda há valores faltantes nas colunas imputadas
print("Valores faltantes após imputação KNN:")
print(dados[variaveis].isna().sum())
```

```python
pasta_saida = "C:\_Master\programacao\Guilherme-Betta\IC2\data\out" # Cria uma variável para guardar o caminho de onde o arquivo resultante será salvo

os.makedirs(pasta_saida, exist_ok=True) # Passo necessário para o python salvar o arquivo

nome_arquivo = f"describe()_{datetime.now().strftime('%Y%m%d_%H%M%S')}.xlsx"   # Gera o nome do arquivo de saida com timestamp de quando foi criado

caminho_saida = os.path.join(pasta_saida, nome_arquivo) # Cria o diretório para salvar o arquivo caso ela ainda não exista

dados.to_excel(caminho_saida, engine='openpyxl', index=True)   # Gera o arquivo de saida
```
