```python
import pandas as pd
import numpy as np
import os
from openpyxl import load_workbook
from datetime import datetime
```

```python
### Carregamento do arquivo de interesse

dadosBrutos = pd.read_excel('data\in\dadosBrutos.xlsx') #Faz o python ler o arquivo e transformá-lo num DataFrame
```

```python
### Impedir que o pandas registre o horário na coluna de datas
dadosBrutos['data_normalizada'] = dadosBrutos['Data'].dt.date   # Cria uma coluna que extrai apenas ano-mes-dia da coluna "Data"

col = dadosBrutos.pop('data_normalizada')   # Extrai o value da coluna criada para a linha seguinte

dadosBrutos.insert(1, 'data_normalizada', col)  # Move a coluna criada para o começo da tabela
```

```python
### Converter Coluna de Estatística em 3 colunas para cada variável

variaveis = [col for col in dadosBrutos.columns if col not in ['Data', 'data_normalizada', 'Estatistica']]   # Separa as colunas de variaveis (que serao alteradas) das colunas de data e estatística

dadosBrutos = dadosBrutos.pivot(index=["Data", "data_normalizada"], columns='Estatistica', values=variaveis)  # Mantem a coluna de datas como indice, converte a coluna de estatiticas em 3 colunas para cada variavel

dadosBrutos.columns = [f"{var}_{est}" for var, est in dadosBrutos.columns]  # A linha acima criou uma linha adicional com as estatisticas, repetindo 3 colunas para/
                                                                            # cada variavel. Essa linha combina a linha de variaveis com a de estatisticas, e as nomeia/
                                                                            # combinando o nome da variavel com o nome da estatistica, mas separando cada nome por um "_"/
                                                                            # Exemplo: "pH" + "Max" = "pH_Max"
```

```python
### Criar um arquivo de saida 

pasta_saida = "data\\out" # Cria uma variável para guardar o caminho de onde o arquivo resultante será salvo

os.makedirs(pasta_saida, exist_ok=True)   # Cria um diretorio para salvar o arquivo caso ele ainda nao exista

nome_arquivo = f"dadosBrutosProntos_{datetime.now().strftime('%Y%m%d_%H%M%S')}.xlsx" # Gera o nome do arquivo de saida com timestamp de quando foi criado

caminho_saida = os.path.join(pasta_saida, nome_arquivo)  # Cria o diretório para salvar o arquivo caso ela ainda não exista

dadosBrutos.to_excel(caminho_saida, engine='openpyxl', index=True)  # Gera o arquivo de saida

print(f"Arquivo salvo em: {caminho_saida}") # Gera uma mensagem para o usuario informando sobre onde o arquivo foi salvo
```
