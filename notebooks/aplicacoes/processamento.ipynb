# Limpeza dos dados
Gera `data/processed/dados_limpos.xlsx`, a base lida pelos notebooks de aplicações, a partir de `data/processed/brutos_prontos.xlsx`:
- converte as variáveis para número;
- tira as linhas sem nenhum valor (2024-12 está vazio no PDF do SEMAE);
- tira as variáveis com mais de 30% de dados faltantes.

```python
import pandas as pd
from pcj.utils import BASE_DIR
from pcj.processamento import limpeza
```

```python
# A primeira coluna do arquivo é o número da linha: vira o índice
brutos = pd.read_excel(BASE_DIR / "data" / "processed" / "brutos_prontos.xlsx", index_col=0)

df_limpo, metadados, variaveis = limpeza(brutos, salvar_arquivo=False)
print("Brutos:", brutos.shape, "| limpos:", df_limpo.shape, "| variáveis:", len(variaveis))
```

Linhas tiradas por não terem nenhum valor

```python
brutos.loc[~brutos.index.isin(df_limpo.index), ["Data", "Estatistica"]]
```

Gravação. A `salvar()` grava em `data/output`, com data e hora no nome; esta base precisa ficar em `data/processed`, com nome fixo. O índice (número da linha nos brutos) vai para a coluna `Index`, como no arquivo anterior.

```python
df_limpo.to_excel(BASE_DIR / "data" / "processed" / "dados_limpos.xlsx", index=True, index_label="Index")
```
